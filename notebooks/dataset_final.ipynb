# Dataset Final

```python
from pathlib import Path
import hashlib
import shutil
import openpyxl

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SUMBER = ROOT / "data" / "Dataset GelarRasa Bismillah aja (3).xlsx"
OUTPUT = ROOT / "hasil" / "dataset_final.xlsx"
OUTPUT.parent.mkdir(parents=True, exist_ok=True)
if not SUMBER.is_file():
    raise FileNotFoundError(f"File belum ada: {SUMBER}")
```

## Simpan hasil

```python
shutil.copyfile(SUMBER, OUTPUT)
def sha256(path):
    h = hashlib.sha256()
    with path.open("rb") as f:
        for blok in iter(lambda: f.read(1024 * 1024), b""):
            h.update(blok)
    return h.hexdigest()

assert sha256(SUMBER) == sha256(OUTPUT)
print(f"Tersimpan: {OUTPUT.name}")
```

## Lihat sheet

```python
wb = openpyxl.load_workbook(OUTPUT, read_only=True, data_only=True)
for nama in wb.sheetnames:
    print(nama)
wb.close()
```
